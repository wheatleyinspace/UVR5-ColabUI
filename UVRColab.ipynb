```python
#@markdown Choose language here.
#@markdown Выберите здесь язык.
language = "ru_RU"  # @param ["ar_AR","de_DE","en_US","es_ES","fr-FR","hi_IN","id_ID","it_IT","ja_JP","ko_KR","ms_MY","pt_BR","ru_RU","th_TH","tr_TR","uk_UA","vi_VN","zh_CN"]

import json
import os

config_path = "./assets/config.json"

if os.path.exists(config_path):
    with open(config_path, "r", encoding="utf-8") as f:
        config = json.load(f)
else:
    config = {
        "theme": {"file": None, "class": "NoCrypt/miku"},
        "lang": {"override": False, "selected_lang": "ru_RU"},
        "discord_presence": True,
        "load_custom_settings": False
    }

# Обновляем язык
config["lang"]["selected_lang"] = language
config["lang"]["override"] = True

with open(config_path, "w", encoding="utf-8") as f:
    json.dump(config, f, indent=2, ensure_ascii=False)
print(f"Language successfully installed: {language}")
print(f"Язык успешно установлен: {language}")
```
